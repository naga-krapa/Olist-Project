```python
import pandas as pd
import os
import matplotlib.pyplot as plt
folder_path = "C:/Users/nagal/OneDrive/Documents/Learn/Repo/Olist-Project/data"
data = {}
for file in os.listdir(folder_path):
    # print(file.split(".")[0])
    data[file.split(".")[0].replace("olist_","").replace("_dataset","").replace("order_","")] = folder_path + "/" + file
data
    
```

```python
38. What is total revenue?
order_items----> sum of all prices = total_revenue

df_items = pd.read_csv(data["items"])
total_revenue = df_items["price"].sum()
print("total_revenue:",total_revenue)

“I calculated total revenue by summing the price column from the Order Items table, 
because each row represents an item sold. The total revenue is approximately $13.59 million.”
```

```python
df_items = pd.read_csv(data["items"])
df_items.columns
df_items["order_id"].is_unique
total_revenue = df_items["price"].sum()
print("total_revenue:",total_revenue)
df_items.shape
df_items["order_id"].nunique()
df = df_items.groupby("order_id")["price"].sum().sort_values(ascending=False).reset_index(name="count")
print("revenue_after_merge:",df["count"].sum())
```

```python
39. How many orders were placed?

orders -------> order_id ------------> count of order_id is total orders placed  
“I calculated the total number of orders by counting the unique order_id values in the Orders table.
    There were 99,441 orders.”
```

```python
df_orders = pd.read_csv(data["orders"])
total_orders = df_orders["order_id"].nunique()
print("total_orders :",total_orders)
```

```python
40. How many unique customers are there?
customers -------> customer_id ----------> number of unique customers
“I calculated the number of unique customers by using nunique() on the customer_id column from the Customers table. 
    There are 99,441 unique customers.”

customer_id and customer_unique_id are different.
If the interviewer asks “How many unique customers placed orders?”, 
we should discuss customer_unique_id, because the same customer can have multiple customer_id values across orders.
```

```python
df_customers = pd.read_csv(data["customers"])
unique_customers = df_customers["customer_id"].nunique()
print("unique_customers:",unique_customers)
```

```python
41. What is the average order value?
items ------> order_id (groupby) ---------> price[sum] ------> mean = average  

“I calculated the average order value by grouping the Order Items by order_id and summing the price for each order.
    Then I calculated the mean of those order-level values.”
```

```python
df_items = pd.read_csv(data["items"])
order_value = df_items.groupby("order_id")["price"].sum()
average_order_value = order_value.mean()
print("average_order_value :",average_order_value )
```

```python
42. What is the average number of items per order?
items ------> order_id (groupby) ,order_item_id(count) -----> order_item_id (avg)

“I grouped the Order Items by order_id and counted the order_item_id values to get the number of items in each order. 
    Then I calculated the mean of those values to get the average number of items per order.”
```

```python
df = df_items.groupby("order_id")["order_item_id"].count().sort_values(ascending=False).reset_index(name="items_per_order")
average_items_per_order = df["items_per_order"].mean()
print("average_items_per_order:",average_items_per_order)
```

```python
43. How many sellers are there?
sellers ---------> seller_id ---------> nunique = total number of sellers
“I calculated the total number of sellers by counting the unique seller_id values in the Sellers table. 
    There are 3,095 unique sellers.”
```
