```python
37. Which tables should be aggregated before merging?
I should aggregate the tables before merging when there are multiple rows for the same `order_id`.
In the Olist dataset, I grouped the Order Items by `order_id` and summed the `price`. 
I also grouped the Payments by `order_id` and summed the `payment_value`.
This gives me one row per order in both tables. Then I can merge them using `order_id` without creating duplicate or extra rows.
This helps prevent revenue or payment values from being counted multiple times.
```
