```python
def clean_data(orders,reviews,items):
    
    order_date_cols = ['order_purchase_timestamp',
                       'order_approved_at', 
                       'order_delivered_carrier_date',
                       'order_delivered_customer_date', 
                       'order_estimated_delivery_date']
    for col in order_date_cols:
        orders[col] = pd.to_datetime(orders[col])
    
    review_date_cols = ['review_creation_date',
                        'review_answer_timestamp']
    for col in review_date_cols:
        reviews[col] = pd.to_datetime(reviews[col])
    
    item_date_cols = ['shipping_limit_date']
    for col in item_date_cols:
        items[col] = pd.to_datetime(items[col])
    return (orders,reviews,items)
```
