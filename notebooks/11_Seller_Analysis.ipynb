```python
63. Which sellers have higher review scores?
I merged the reviews and items tables using order ID. 
Then I grouped the data by seller and calculated both the average review score and review count. 
I filtered out sellers with fewer than 50 reviews to make the comparison more strong, 
and finally sorted the sellers by average review score in descending order.

    seller_id = "d9bd94811c3338dceb4181f3dbc0c73e",review_mean = 4.819672, count = 61,city = sao paulo,state=SP
```

```python
df_reviews = pd.read_csv(data["reviews"])
reviews_items = pd.merge(df_reviews,df_items,on="order_id",how="inner")
#score = reviews_items.groupby("seller_id")["review_score"].mean().sort_values(ascending=False).reset_index(name="review_score")
seller_score = (reviews_items.groupby("seller_id")["review_score"].agg(["mean", "count"]).reset_index())

seller_score = seller_score[seller_score["count"] >= 50].sort_values("mean", ascending=False)
```
