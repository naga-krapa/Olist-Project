```python
51. Which categories have the highest average order value?

Average order value → first calculate total value per order, then calculate the average of those order values.
```

```python
products_items = pd.merge(
    df_products,
    df_items,
    on="product_id",
    how="inner"
)

order_value_category = (
    products_items
    .groupby(["product_category_name", "order_id"])["price"]
    .sum()
    .reset_index(name="order_value")
)

average_order_value_category = (
    order_value_category
    .groupby("product_category_name")["order_value"]
    .mean()
    .sort_values(ascending=False)
)

average_order_value_category
```

```python
53. Which categories have the best review scores?
Merged items with products
Merged with reviews
Grouped by product_category_name
Calculated the average review score using .mean()
Sorted ascending to False get the highest scores first
Merged with the English category translation
cds_dvds_musicals ----------------> 4.642857 using round(2) -------> 4.64
```

```python
df_reviews = pd.read_csv(data["reviews"])
df_product_category_name_translation = pd.read_csv(data["product_category_name_translation"])
items_products = pd.merge(df_items,df_products,on="product_id",how="inner")
items_products_reviews = pd.merge(items_products,df_reviews,on="order_id",how="inner")
avg_review_score = items_products_reviews.groupby("product_category_name")["review_score"].mean().sort_values(ascending=False).reset_index(name="average_score")
#pd.merge(max_review_score,df_product_category_name_translation,on="product_category_name",how="inner")
print("average maximum product category score:",avg_review_score.iloc[0,0],"with",avg_review_score.iloc[0,1],"maximum score")
print("average minimum product category score:",avg_review_score.iloc[-1,0],"with",avg_review_score.iloc[-1,1],"minimum score")
```
