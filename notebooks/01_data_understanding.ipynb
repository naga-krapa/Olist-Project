```python
#  5. What are the primary keys?
# in customers table {customer_id }  --------->  Unique: True  and   Nulls: 0  -------------->   PK
# in orders table  {order_id}  --------------->  Unique: True  and   Nulls: 0  -------------->   PK
#                  {customer_id}  ------------>  Unique: True  and   Nulls: 0  -------------->   FK
# in products table{product_id}  ------------->  Unique: True  and   Nulls: 0  -------------->   PK
# in item table   {order_id}   --------------->  Unique: False and   Nulls: 0  ----------->     FK  b/c it connects items---->orders
#                 {order_item_id}------------->  Unique: False and   Nulls: 0  ------------>  FK b/c it connects items--->products

# in sellers table {seller_id}  -------------->  Unique: True  and   Nulls: 0  --->  PK
# in product_category_name_translation  ------> { product_category_name}      Unique: True   and  Nulls: 0
#                                          {product_category_name_english}    Unique: True   and  Nulls: 0

```

```python
# 5. What are the primary keys and foreign keys?

customers  -------------------->  customer_id → PK

orders  ---------------------->  order_id  → PK

       ----------------------->  customer_id → FK → customers.customer_id


products ----------------------> product_id  → PK


sellers  ---------------------->  seller_id  → PK

product_category_name_translation  -------------->  product_category_name
                                                 → Key used to connect to products.product_category_name
order_items
    (order_id, order_item_id)
    → Composite/natural key

    order_id
    → FK → orders.order_id

    product_id
    → FK → products.product_id

    seller_id
    → FK → sellers.seller_id




```

```python
# 6. Which columns connect the tables?
'Customers' + 'orders'      =  "customer_id"
'customers' + 'geolocation' = "ZIP prefix"
'sellers'   + 'items'       = "seller_id"
'sellers'   + 'geolocation' = "zip_prefix"
'products'  + 'items'       =  "product_id"
'products'  + 'product_category_name_translation' = "product_category_name"
'orders'    + 'items'       = "order_id"
'orders'    + 'reviews'     =  "order_id"
'payments'  +  'orders'     =  "order_id"

```
