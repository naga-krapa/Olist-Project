```python
Overall conclusion for Q27–Q32

Overall, the Olist dataset has a relational structure centered around orders, 
with order_id, customer_id, product_id, and seller_id connecting the major tables. 
Orders can contain multiple payment records and multiple order items, 
creating one-to-many relationships with the Payments and Order Items tables.

The customer analysis shows that while a customer_id is generally associated with one order record, 
using customer_unique_id reveals repeat purchasing behavior, with 2,997 unique customers placing more than one order. 
This distinction is important because customer_id and customer_unique_id represent different levels of customer identity.

Order Items provides the detailed product-level transaction information. 
It connects orders with products and sellers, allowing us to analyze product demand and seller activity. 
The Products → Order Items and Sellers → Order Items relationships are one-to-many 
because a single product or seller can appear in many order-item records.

The product analysis shows that 32,951 products were ordered at least once, 
while 14,834 products appeared in more than one order-item record. 
The most frequently appearing product had 527 order-item records. 
Similarly, seller activity varies considerably: some sellers have many order-item records while others have only a small number.

For revenue analysis, the Order Items table is an appropriate starting point 
because it contains product-level price and freight_value information. 
It can then be joined with Orders, Products, Customers, Sellers, and Payments depending on the business question.

Overall, these relationships provide the foundation for analyzing customer behavior, repeat purchases, product demand, seller performance, payment behavior, and revenue.
```
